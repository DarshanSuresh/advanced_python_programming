## 01) Printing Nested Dictionary for Semester X printing wiht name and marks?

```python
semester_2_students = {
    1: {"name": "Alice", "marks": {"C++": 85}},
    2: {"name": "Bob", "marks": {"Python": 78}},
    3: {"name": "Charlie", "marks": {"AI": 92}},
    4: {"name": "Darshan", "marks": {"DBMS": 97}}
}

print("--- Semester 2 Student Records ---")
for student_id, details in semester_2_students.main() if hasattr(semester_2_students, 'main') else semester_2_students.items():
    name = details["name"]
    marks = details["marks"]

    print(f"\nStudent Name: {name}")
    print("Marks:")
    for subject, score in marks.items():
        print(f"  - {subject}: {score}")
```

## 02) Palindrome of numbers?

```python
def is_palindrome_math(number):
    if number < 0:
        return False
    original = number
    reversed_num = 0
    while number > 0:
        digit = number % 10
        reversed_num = (reversed_num * 10) + digit
        number //= 10
    return original == reversed_num
print(is_palindrome_math(232))
```

## 01) Find the maximum value from a collection of argument using *args (positional arbitrary)

```python
def find_maximum(*args):
    if not args:
        return "No arguments provided"
    max_val = args[0]
    for num in args[1:]:
        if num > max_val:
            max_val = num
    return max_val

print(find_maximum(3, 2, 1, 9, 11))
```

## 02) Sumation of all the arguments?

```python
def find_sum(*args):
    if not args:
        return 0
    total = 0
    for num in args:
        total += num
    return total
print(find_sum(3, 2, 1, 9, 11))
```
